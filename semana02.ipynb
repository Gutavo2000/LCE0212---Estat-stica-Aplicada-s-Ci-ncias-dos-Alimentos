```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Excel file into a pandas DataFrame (if not already loaded)
df = pd.read_excel('/content/dados_semana2.xlsx')

# Histogram for 'Salário Mensal (R$) (Contínua)'
plt.figure(figsize=(10, 6))
sns.histplot(df['Salário Mensal (R$) (Contínua)'], kde=True)
plt.title('Distribuição de Salário Mensal')
plt.xlabel('Salário Mensal (R$)')
plt.ylabel('Frequência')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Excel file into a pandas DataFrame (if not already loaded)
df = pd.read_excel('/content/dados_semana2.xlsx')

# Bar chart for 'Projetos Concluídos (Discreta)'
plt.figure(figsize=(10, 6))
sns.countplot(x='Projetos Concluídos (Discreta)', data=df, palette='viridis')
plt.title('Contagem de Projetos Concluídos')
plt.xlabel('Número de Projetos Concluídos')
plt.ylabel('Contagem')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Excel file into a pandas DataFrame (if not already loaded)
df = pd.read_excel('/content/dados_semana2.xlsx')

# Pie chart for 'Departamento (Nominal)'
department_counts = df['Departamento (Nominal)'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(department_counts, labels=department_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Distribuição de Funcionários por Departamento')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```
