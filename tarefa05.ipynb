## Carregando os Dados

Este passo carrega o arquivo `Tarefa05.xlsx` em um DataFrame do pandas para que possamos começar a trabalhar com os dados.

```python
import pandas as pd

# Carregar o arquivo Excel
df = pd.read_excel('/content/Tafera05.xlsx')

# Exibir as primeiras 5 linhas para entender a estrutura dos dados
display(df.head())
```

## Análise Exploratória: Gráfico Boxplot de Proteína por Grupo

Este código gera um gráfico boxplot para visualizar a distribuição do teor de proteína em cada grupo. Ele nos ajuda a identificar a mediana, a dispersão e possíveis outliers do teor de proteína para cada categoria.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Criar o gráfico boxplot
plt.figure(figsize=(12, 7))
sns.boxplot(x='Categoria', y='Proteína (g)', data=df, palette='viridis', hue='Categoria', legend=False)

# Adicionar título e rótulos
plt.title('Teor de Proteína por Categoria', fontsize=16)
plt.xlabel('Categoria', fontsize=12)
plt.ylabel('Teor de Proteína (g)', fontsize=12)
plt.xticks(rotation=45, ha='right') # Rotacionar os rótulos do eixo X para melhor visualização
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout() # Ajusta o layout para evitar sobreposição
plt.show()
```

## Estatísticas Descritivas do Teor de Proteína por Categoria

Este código calcula a média, mediana e desvio-padrão do teor de 'Proteína (g)' para cada 'Categoria' de alimento. Essas estatísticas complementam o boxplot anterior, oferecendo medidas quantitativas de tendência central e dispersão.

```python
# Calcular as estatísticas descritivas para Proteína (g) por categoria
protein_stats = df.groupby('Categoria')['Proteína (g)'].agg([
    'mean',   # Média
    'median', # Mediana
    'std'     # Desvio-padrão
])

# Exibir as estatísticas descritivas do teor de proteína
display(protein_stats)
```

##Os diferentes grupos de alimentos apresentam comportamentos semelhantes quanto ao teor de proteína? Como os dados podem ser utilizados para comparar esses grupos?

Os diferentes grupos de alimentos não apresentam comportamentos semelhantes quanto ao teor de proteína. Observa-se que carnes bovinas, carnes de aves e laticínios e ovos apresentam os maiores teores médios de proteína, enquanto frutas e vegetais apresentam os menores valores. Para comparar os grupos, podem ser utilizadas medidas estatísticas como média, mediana, quartis e desvio-padrão, juntamente com o gráfico de boxplot. Essas informações permitem comparar a tendência central e a variabilidade do teor de proteína entre as diferentes categorias de alimentos.

## Análise Comparativa: Proteína, Gordura Total e Sódio por Categoria

Este código gera três gráficos boxplot em uma única figura para comparar a distribuição de Proteína, Gordura Total e Sódio entre as diferentes categorias de alimentos. A visualização lado a lado facilita a identificação de padrões e diferenças nas distribuições dessas métricas nutricionais.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Criar uma figura com 1 linha e 3 colunas para os subplots
fig, axes = plt.subplots(1, 3, figsize=(24, 8))

# Boxplot para Proteína
sns.boxplot(x='Categoria', y='Proteína (g)', data=df, palette='viridis', hue='Categoria', legend=False, ax=axes[0])
axes[0].set_title('Teor de Proteína (g) por Categoria', fontsize=14)
axes[0].set_xlabel('Categoria', fontsize=12)
axes[0].set_ylabel('Proteína (g)', fontsize=12)
axes[0].tick_params(axis='x', rotation=45)
axes[0].grid(axis='y', linestyle='--', alpha=0.7)

# Boxplot para Gordura Total
sns.boxplot(x='Categoria', y='Gordura total (g)', data=df, palette='magma', hue='Categoria', legend=False, ax=axes[1])
axes[1].set_title('Teor de Gordura Total (g) por Categoria', fontsize=14)
axes[1].set_xlabel('Categoria', fontsize=12)
axes[1].set_ylabel('Gordura total (g)', fontsize=12)
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(axis='y', linestyle='--', alpha=0.7)

# Boxplot para Sódio
sns.boxplot(x='Categoria', y='Sódio (mg)', data=df, palette='cividis', hue='Categoria', legend=False, ax=axes[2])
axes[2].set_title('Teor de Sódio (mg) por Categoria', fontsize=14)
axes[2].set_xlabel('Categoria', fontsize=12)
axes[2].set_ylabel('Sódio (mg)', fontsize=12)
axes[2].tick_params(axis='x', rotation=45)
axes[2].grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout() # Ajusta o layout para evitar sobreposição
plt.show()
```

## Análise Estatística Descritiva por Categoria

Este código calcula as estatísticas descritivas (média, mediana e desvio-padrão) para os teores de Proteína, Gordura Total e Sódio, agrupados por Categoria de Alimento. Isso oferece um resumo quantitativo da distribuição desses nutrientes em cada grupo.

```python
import pandas as pd

# Definir as colunas de nutrientes para as quais queremos calcular as estatísticas
nutrient_columns = ['Proteína (g)', 'Gordura total (g)', 'Sódio (mg)']

# Calcular as estatísticas descritivas por categoria
df_stats = df.groupby('Categoria')[nutrient_columns].agg([
    'mean',   # Média
    'median', # Mediana
    'std'     # Desvio-padrão
])

# Exibir as estatísticas descritivas
display(df_stats)
```

## Após as análises realizadas, quais características estatísticas diferenciam mais claramente os grupos de alimentos?

Com base nos graficos que fizemos de boxsplot e de analise estatistica descritiva podemos afirmar que as caracteristicas que mais se diferenciam em todos os grupos de alimentos são a mediana que resultou em valores distintos entre os analisados principalemte no teor de proteinas. Outro indice marcante seria o intervalo interquartil que permite identificar diferenças de variabilidade dos valores dentro de cada grupo.
