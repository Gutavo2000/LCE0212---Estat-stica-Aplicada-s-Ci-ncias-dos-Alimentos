## Introdução

No universo da produção de alimentos, a eficiência e a qualidade são cruciais. No entanto, falhas operacionais podem ocorrer, impactando a produtividade e a segurança. Compreender a frequência e a probabilidade dessas falhas é fundamental para otimizar os processos e implementar medidas preventivas. Este estudo se propõe a analisar uma situação-problema comum em linhas de processamento: a ocorrência de falhas operacionais. Utilizaremos conceitos de probabilidade, especificamente a Distribuição de Poisson, para modelar e prever o número de falhas em determinados períodos, fornecendo insights valiosos para a gestão da qualidade na indústria de alimentos.

## Definição da Variável Aleatória

Seja **X** a variável aleatória que representa o **número de falhas operacionais registradas durante um determinado período de tempo** em uma etapa da linha de processamento de alimentos.

### Questão 1: Quais valores a variável aleatória X pode assumir?

A variável aleatória X, que representa o número de falhas, pode assumir apenas valores inteiros não negativos. Isso porque não podemos ter um número fracionado de falhas nem um número negativo de falhas. Portanto, X pode assumir os valores: 0, 1, 2, 3, ... (e assim por diante, infinitamente).

## Explicação de Por Que a Distribuição de Poisson é Adequada

A Distribuição de Poisson é um modelo de probabilidade discreto que é frequentemente utilizado para modelar o número de ocorrências de um evento em um intervalo fixo de tempo ou espaço, quando esses eventos acontecem com uma taxa média conhecida e independentemente uns dos outros. Para a nossa situação-problema, a Distribuição de Poisson é adequada por várias razões:

1.  **Eventos Raros**: As falhas são eventos que, embora possam acontecer, são geralmente considerados 'raros' ou com uma baixa probabilidade de ocorrer em um curto espaço de tempo.
2.  **Independência**: As ocorrências das falhas são independentes entre si, ou seja, uma falha não afeta a probabilidade da próxima falha.
3.  **Taxa Média Constante**: A taxa média de ocorrência das falhas (0,4 falhas por hora) é constante ao longo do período analisado.
4.  **Variável Discreta**: A variável X (número de falhas) é uma variável discreta, que assume valores inteiros (0, 1, 2, ...).
5.  **Intervalo Fixo**: Estamos observando as falhas dentro de um intervalo fixo de tempo (uma hora ou três horas).

Essas características se alinham perfeitamente aos pressupostos da Distribuição de Poisson, tornando-a uma ferramenta estatística robusta para analisar nosso problema.

## Apresentação da Fórmula da Distribuição de Poisson

A função de probabilidade de massa (PMF) da Distribuição de Poisson é dada por:

$P(X = k) = \frac{e^{-\lambda} \lambda^k}{k!}$

Onde:
*   $P(X = k)$ é a probabilidade de que $X$ seja igual a um número específico de ocorrências $k$.
*   $e$ é a base do logaritmo natural (aproximadamente 2.71828).
*   $\lambda$ (lambda) é a taxa média de ocorrências no intervalo de tempo/espaço. É o único parâmetro da distribuição.
*   $k$ é o número de ocorrências desejado (0, 1, 2, ...).
*   $k!$ é o fatorial de $k$ ($k \times (k-1) \times ... \times 1$).

### Questão 2: Qual é a distribuição de probabilidade de X? Identifique seu parâmetro.

A distribuição de probabilidade de X é a **Distribuição de Poisson**. Seu único parâmetro é $\lambda$ (lambda), que representa a taxa média de ocorrências no período de observação. Para uma hora, $\lambda = 0,4$.

```python
# Importando as bibliotecas necessárias
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
import pandas as pd # Para a tabela resumo
```

## Resolução das Questões

### Questão 3: Qual é a probabilidade de nenhuma falha ser registrada durante uma hora?

Para calcular a probabilidade de nenhuma falha (X = 0) em uma hora, onde $\lambda = 0,4$, aplicamos a fórmula da Distribuição de Poisson:

$P(X = 0) = \frac{e^{-0,4} (0,4)^0}{0!} = \frac{e^{-0,4} \times 1}{1} = e^{-0,4}$

```python
# Parâmetro lambda para uma hora
lambda_1h = 0.4

# Cálculo da probabilidade de X = 0 usando a função pmf (probability mass function) da distribuição de Poisson
prob_0_falhas_1h = poisson.pmf(0, lambda_1h)

print(f"Probabilidade de nenhuma falha em 1 hora (decimal): {prob_0_falhas_1h:.4f}")
print(f"Probabilidade de nenhuma falha em 1 hora (porcentagem): {prob_0_falhas_1h * 100:.2f}%")
```

### Interpretação do Resultado (Questão 3)

Com uma taxa média de 0,4 falhas por hora, há aproximadamente **67,03% de chance** de que nenhuma falha seja registrada na linha de processamento durante uma hora de observação. Isso indica que, na maioria das vezes, a linha opera sem intercorrências de falha durante esse período.

### Questão 4: Qual é a probabilidade de pelo menos uma falha ser registrada durante uma hora?

A probabilidade de pelo menos uma falha (X $\ge$ 1) é o complemento da probabilidade de nenhuma falha (X = 0). Isso significa que:

$P(X \ge 1) = 1 - P(X = 0)$

Já calculamos $P(X = 0)$ na questão anterior.

```python
# Cálculo da probabilidade de pelo menos uma falha (X >= 1) usando o evento complementar
prob_pelo_menos_1_falha_1h = 1 - prob_0_falhas_1h

print(f"Probabilidade de pelo menos uma falha em 1 hora (decimal): {prob_pelo_menos_1_falha_1h:.4f}")
print(f"Probabilidade de pelo menos uma falha em 1 hora (porcentagem): {prob_pelo_menos_1_falha_1h * 100:.2f}%")
```

### Interpretação do Resultado (Questão 4)

Existe uma probabilidade de aproximadamente **32,97%** de que pelo menos uma falha ocorra na linha de processamento dentro de uma hora. Embora a probabilidade de **nenhuma** falha seja maior, a chance de uma falha acontecer não é desprezível e merece atenção para garantir a continuidade da produção.

## Análise para um Período de Três Horas

Agora, vamos considerar que a linha de produção seja observada durante um período maior: três horas.

### Questão 5: Qual deve ser o valor do parâmetro λ para esse novo período?

Como a taxa média de falhas é de 0,4 falhas por hora e o período de observação é de três horas, o novo valor de $\lambda$ será a taxa média multiplicada pelo número de horas:

$\lambda_{\text{3h}} = \text{taxa média por hora} \times \text{número de horas}$
$\lambda_{\text{3h}} = 0,4 \times 3 = 1,2$

```python
# Cálculo do novo parâmetro lambda para três horas
lambda_3h = 0.4 * 3

print(f"O novo valor do parâmetro lambda para 3 horas é: {lambda_3h:.1f}")
```

### Questão 6: Qual é a probabilidade de nenhuma falha ser registrada durante as três horas?

Para calcular a probabilidade de nenhuma falha (X = 0) em três horas, onde $\lambda = 1,2$, aplicamos a fórmula da Distribuição de Poisson:

$P(X = 0) = \frac{e^{-1,2} (1,2)^0}{0!} = \frac{e^{-1,2} \times 1}{1} = e^{-1,2}$

```python
# Cálculo da probabilidade de X = 0 para 3 horas
prob_0_falhas_3h = poisson.pmf(0, lambda_3h)

print(f"Probabilidade de nenhuma falha em 3 horas (decimal): {prob_0_falhas_3h:.4f}")
print(f"Probabilidade de nenhuma falha em 3 horas (porcentagem): {prob_0_falhas_3h * 100:.2f}%")
```

### Interpretação do Resultado (Questão 6)

Ao observar a linha por três horas, a probabilidade de **nenhuma** falha ser registrada diminui significativamente para aproximadamente **30,12%**. Isso é esperado, pois quanto maior o período de observação, maior a chance de que pelo menos uma falha ocorra, dado que a taxa média por hora é constante.

### Questão 7: Explique por que o valor de λ foi alterado quando o período de observação passou de uma para três horas, embora o modelo probabilístico continue sendo o mesmo.

O valor do parâmetro $\lambda$ (lambda) foi alterado de 0,4 para 1,2 porque $\lambda$ representa a **taxa média esperada de ocorrências no intervalo de tempo sob análise**. Embora a taxa média de falhas *por hora* continue sendo a mesma (0,4 falhas/hora), o *período total de observação* mudou de uma hora para três horas.

Se esperamos 0,4 falhas em uma hora, é intuitivo esperar o triplo de falhas em três horas, mantendo a mesma taxa. A Distribuição de Poisson modela o número de eventos em um intervalo, e esse intervalo é parte integrante de seu parâmetro $\lambda$. A natureza do evento (falha operacional) e suas propriedades (independência, taxa constante) permanecem as mesmas, por isso o **modelo probabilístico (Poisson) continua sendo o mesmo**, mas seu parâmetro se ajusta ao novo tamanho do intervalo de observação.

## Visualização da Distribuição de Poisson

Para entender melhor como a probabilidade das falhas se distribui, vamos gerar gráficos da Distribuição de Poisson para ambos os cenários (1 hora e 3 horas). Isso nos ajudará a visualizar como a probabilidade de diferentes números de falhas varia com o parâmetro $\lambda$.

```python
# Definindo os valores de X para os quais queremos calcular a probabilidade
x_values_1h = np.arange(0, 6) # De 0 a 5 falhas

# Calculando as probabilidades para lambda = 0.4 (1 hora)
probabilities_1h = poisson.pmf(x_values_1h, lambda_1h)

# Criando o gráfico para 1 hora
plt.figure(figsize=(10, 6))
plt.bar(x_values_1h, probabilities_1h, color='skyblue', edgecolor='black')

# Adicionando rótulos e título
plt.title(f'Distribuição de Poisson para 1 Hora (λ = {lambda_1h})')
plt.xlabel('Número de Falhas (X)')
plt.ylabel('Probabilidade P(X=x)')
plt.xticks(x_values_1h) # Garante que todos os valores de X apareçam no eixo
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Adicionando os valores de probabilidade em cima das barras
for i, prob in enumerate(probabilities_1h):
    plt.text(x_values_1h[i], prob + 0.01, f'{prob:.3f}', ha='center', va='bottom', fontsize=9)

plt.show()
```

```python
# Definindo os valores de X para os quais queremos calcular a probabilidade para 3 horas
# Para lambda = 1.2, é bom estender um pouco mais para ver o pico da distribuição
x_values_3h = np.arange(0, 8) # De 0 a 7 falhas

# Calculando as probabilidades para lambda = 1.2 (3 horas)
probabilities_3h = poisson.pmf(x_values_3h, lambda_3h)

# Criando o gráfico para 3 horas
plt.figure(figsize=(10, 6))
plt.bar(x_values_3h, probabilities_3h, color='lightcoral', edgecolor='black')

# Adicionando rótulos e título
plt.title(f'Distribuição de Poisson para 3 Horas (λ = {lambda_3h})')
plt.xlabel('Número de Falhas (X)')
plt.ylabel('Probabilidade P(X=x)')
plt.xticks(x_values_3h) # Garante que todos os valores de X apareçam no eixo
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Adicionando os valores de probabilidade em cima das barras
for i, prob in enumerate(probabilities_3h):
    plt.text(x_values_3h[i], prob + 0.005, f'{prob:.3f}', ha='center', va='bottom', fontsize=9)

plt.show()
```

### Interpretação e Comparação dos Gráficos

Ao comparar os dois gráficos de Distribuição de Poisson, podemos observar diferenças claras que refletem o aumento do período de observação:

1.  **Distribuição para 1 Hora ($\lambda = 0,4$)**: O pico de probabilidade está em 0 falhas, indicando que é o evento mais provável. As probabilidades diminuem rapidamente à medida que o número de falhas aumenta, sendo muito baixa a chance de 3 ou mais falhas.

2.  **Distribuição para 3 Horas ($\lambda = 1,2$)**: O pico da distribuição se desloca para a direita, agora em 1 falha. Isso significa que, em três horas, a probabilidade de ocorrer **uma falha** é maior do que a probabilidade de não ocorrer nenhuma falha. A distribuição se espalha mais, mostrando probabilidades consideráveis para 0, 1, 2 ou até 3 falhas.

**Por que o aumento do tempo de observação aumenta a quantidade média esperada de falhas?**

Isso ocorre porque a Distribuição de Poisson é baseada em uma **taxa de ocorrência constante por unidade de tempo**. Se você estender o período de observação, estará simplesmente fornecendo mais 'oportunidades' para as falhas acontecerem, mantendo a mesma taxa por hora. Consequentemente, o número médio de falhas esperado (o próprio $\lambda$) no intervalo *total* aumenta proporcionalmente ao tempo de observação. Em vez de esperar 0,4 falhas em uma hora, esperamos 1,2 falhas em três horas, o que faz com que a distribuição 'se desloque' para a direita, indicando uma maior probabilidade de observar um número maior de eventos.
