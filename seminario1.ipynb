```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
file_path = '/content/winequality-white.csv'
df = pd.read_csv(file_path, sep=';')

# Display the first 5 rows to understand the data structure
print("First 5 rows of the dataset:")
display(df.head())

print("\nInformation about the dataset:")
df.info()
```

### 1. Distribuição das notas de qualidade — gráfico de barras

```python
# Count the occurrences of each quality score
quality_counts = df['quality'].value_counts().sort_index()

plt.figure(figsize=(10, 6))
sns.barplot(x=quality_counts.index, y=quality_counts.values, palette='viridis')
plt.title('Distribuição das Notas de Qualidade')
plt.xlabel('Qualidade')
plt.ylabel('Contagem')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### 2. Histograma do teor alcoólico

Eixo X: álcool (%)
Eixo Y: frequência

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['alcohol'], kde=True, bins=20, color='skyblue')
plt.title('Histograma do Teor Alcoólico')
plt.xlabel('Álcool (%)')
plt.ylabel('Frequência')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### 3. Boxplot do álcool por categoria de qualidade

Vamos criar três categorias:

*   **Baixa**: qualidade 3–5
*   **Média**: qualidade 6
*   **Alta**: qualidade 7–9

```python
# Create quality categories
def categorize_quality(quality):
    if 3 <= quality <= 5:
        return 'Baixa'
    elif quality == 6:
        return 'Média'
    elif 7 <= quality <= 9:
        return 'Alta'
    return 'Outra'

df['quality_category'] = df['quality'].apply(categorize_quality)

# Ensure the order of categories in the plot
quality_order = ['Baixa', 'Média', 'Alta']
df['quality_category'] = pd.Categorical(df['quality_category'], categories=quality_order, ordered=True)

plt.figure(figsize=(10, 6))
sns.boxplot(x='quality_category', y='alcohol', data=df, palette='pastel')
plt.title('Teor Alcoólico por Categoria de Qualidade')
plt.xlabel('Categoria de Qualidade')
plt.ylabel('Álcool (%)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### 4. Dispersão: álcool × qualidade

Eixo X: teor alcoólico (%)
Eixo Y: qualidade

Cada ponto representa um vinho. Adicionaremos uma linha de tendência.

```python
plt.figure(figsize=(10, 6))
sns.regplot(x='alcohol', y='quality', data=df, scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
plt.title('Dispersão: Álcool vs. Qualidade com Linha de Tendência')
plt.xlabel('Teor Alcoólico (%)')
plt.ylabel('Qualidade')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### 6. Comparando Vinhos Tintos e Brancos

Para comparar os vinhos tinto e branco, primeiro precisamos carregar o conjunto de dados do vinho tinto e combiná-lo com o conjunto de dados do vinho branco. Em seguida, adicionaremos uma nova coluna 'type' para identificar cada tipo de vinho.

```python
# Load the red wine dataset
red_wine_file_path = '/content/winequality-red.csv'
df_red = pd.read_csv(red_wine_file_path, sep=';')

# Add a 'type' column to both dataframes
df['type'] = 'white' # Corrected from df_white to df
df_red['type'] = 'red'

# Combine the two dataframes
df_combined = pd.concat([df, df_red], ignore_index=True)

print("Primeiras 5 linhas do dataset combinado:")
display(df_combined.head())

print("Últimas 5 linhas do dataset combinado:")
display(df_combined.tail())

print("Informações sobre o dataset combinado:")
df_combined.info()
```

### 7. Gráfico de Dispersão Comparativo: pH × Acidez Fixa por Tipo de Vinho

Agora vamos visualizar a relação entre pH e acidez fixa, diferenciando entre vinhos tinto e branco.

```python
plt.figure(figsize=(12, 7))
sns.lmplot(x='pH', y='fixed acidity', hue='type', data=df_combined, scatter_kws={'alpha':0.3}, line_kws={'lw':2})
plt.title('Dispersão: pH vs. Acidez Fixa por Tipo de Vinho com Linhas de Tendência')
plt.xlabel('pH')
plt.ylabel('Acidez Fixa')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### 8. Gráfico de Dispersão Comparativo: Açúcar Residual vs. Densidade por Tipo de Vinho

Agora, vamos explorar a relação entre o açúcar residual e a densidade para vinhos tintos e brancos.

```python
plt.figure(figsize=(12, 7))
sns.lmplot(x='residual sugar', y='density', hue='type', data=df_combined, scatter_kws={'alpha':0.3}, line_kws={'lw':2})
plt.title('Dispersão: Açúcar Residual vs. Densidade por Tipo de Vinho com Linhas de Tendência')
plt.xlabel('Açúcar Residual')
plt.ylabel('Densidade')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### 9. Coeficiente de Correlação: Açúcar Residual vs. Densidade

Vamos calcular o coeficiente de correlação de Pearson entre 'Açúcar Residual' e 'Densidade' para o conjunto de dados combinado, e também para cada tipo de vinho separadamente, para entender melhor a força e a direção da relação linear.

```python
# Coeficiente de correlação para o dataset combinado
correlation_overall = df_combined['residual sugar'].corr(df_combined['density'])
print(f"Coeficiente de Correlação (Geral) entre Açúcar Residual e Densidade: {correlation_overall:.3f}")

# Coeficiente de correlação para vinhos brancos
df_white_filtered = df_combined[df_combined['type'] == 'white']
correlation_white = df_white_filtered['residual sugar'].corr(df_white_filtered['density'])
print(f"Coeficiente de Correlação (Vinho Branco) entre Açúcar Residual e Densidade: {correlation_white:.3f}")

# Coeficiente de correlação para vinhos tintos
df_red_filtered = df_combined[df_combined['type'] == 'red']
correlation_red = df_red_filtered['residual sugar'].corr(df_red_filtered['density'])
print(f"Coeficiente de Correlação (Vinho Tinto) entre Açúcar Residual e Densidade: {correlation_red:.3f}")
```

### 10. Identificando as Variáveis com Maior Correlação Positiva

Vamos calcular a matriz de correlação para todas as variáveis numéricas no `df_combined` para identificar quais pares de variáveis apresentam os coeficientes de correlação mais próximos de 1 (forte correlação positiva).

```python
# Calcular a matriz de correlação para todas as variáveis numéricas
correlation_matrix = df_combined.corr(numeric_only=True)

# Filtrar apenas as correlações positivas (exceto a correlação de uma variável consigo mesma)
positive_correlations = correlation_matrix[correlation_matrix < 1].stack()
positive_correlations = positive_correlations[positive_correlations > 0]

# Ordenar as correlações em ordem decrescente
sorted_correlations = positive_correlations.sort_values(ascending=False)

print("Os 10 maiores coeficientes de correlação positiva (excluindo correlação com a própria variável) são:")
print(sorted_correlations.head(10))
```
