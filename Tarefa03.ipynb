```python
import pandas as pd

# Rodando o arquivo que exportamos no grafico
file_path = '/content/tabela_tarefa03.xlsx'
df = pd.read_excel(file_path)

# Mostrando todas as variaveis que temos na tabela
print("Tabela 03")
display(df)
```

Após carregar os dados, vamos analisar as informações sobre as colunas, como os tipos de dados e a quantidade de valores não nulos, para ajudar a identificar as variáveis adequadas para as tabelas de frequência e os gráficos.

```python
# Exibir informações básicas sobre o DataFrame
print("DataFrame Info:")
df.info()
```

### Frequências para 'Sexo' (Variável Nominal)

```python
# Tabela de frequência para 'Sexo'
freq_sexo = df['Sexo'].value_counts()
rel_freq_sexo = df['Sexo'].value_counts(normalize=True) * 100

freq_table_sexo = pd.DataFrame({
    'Frequência Simples': freq_sexo,
    'Frequência Relativa (%)': rel_freq_sexo
})

print("Tabela de Frequência para 'Sexo':")
display(freq_table_sexo)
```

### Frequências para 'Xicaras_cafe_dia' (Variável Discreta)

```python
# Tabela de frequência para 'Xicaras_cafe_dia'
freq_xicaras = df['Xicaras_cafe_dia'].value_counts().sort_index()
rel_freq_xicaras = df['Xicaras_cafe_dia'].value_counts(normalize=True).sort_index() * 100

freq_table_xicaras = pd.DataFrame({
    'Frequência Simples': freq_xicaras,
    'Frequência Relativa (%)': rel_freq_xicaras
})

print("Tabela de Frequência para 'Xicaras_cafe_dia':")
display(freq_table_xicaras)
```

## Visualizações Gráficas

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Definir o estilo dos gráficos
sns.set_style('whitegrid')

# Gráfico de barras para 'Sexo''
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Sexo', hue='Sexo', palette='viridis', legend=False)
plt.title('Distribuição de Estudantes por Sexo')
plt.xlabel('Sexo')
plt.ylabel('Número de Estudantes')
plt.show()
```

```python
# Histograma para 'Horas_sono
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Horas_sono', bins=5, kde=False, color='skyblue')
plt.title('Distribuição de Horas de Sono por Estudante')
plt.xlabel('Horas de Sono')
plt.ylabel('Frequência')
plt.show()
```

```python
# Gráfico de dispersão de 'Xicaras_cafe_dia' versus 'Gasto_semanal_cafe'
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='Xicaras_cafe_dia', y='Gasto_semanal_cafe', hue='Sexo', s=100, alpha=0.8)
plt.title('Relação entre Xícaras de Café por Dia e Gasto Semanal com Café')
plt.xlabel('Xícaras de Café por Dia')
plt.ylabel('Gasto Semanal com Café (R$)')
plt.legend(title='Sexo')
plt.show()
```
