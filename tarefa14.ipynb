## 2. Contextualização

A distribuição Normal (ou Gaussiana) é um modelo matemático amplamente utilizado em estatística para descrever variáveis quantitativas contínuas. Na Ciência de Alimentos e no controle de qualidade de processos industriais, diversos parâmetros físico-químicos e sensoriais comportam-se como uma curva normal quando o processo está sob controle.

O pH é uma das variáveis mais críticas no monitoramento de processos e na segurança de alimentos, influenciando diretamente a atividade microbiana, estabilidade enzimática, textura e o sabor final do produto. Se o processo produtivo estiver estável, as pequenas variações ambientais, operacionais e de matéria-prima resultam em flutuações simétricas em torno do valor ideal planejado (a média).

**O problema prático:**
Durante o controle de qualidade do produto, o pH pode ser representado por uma distribuição Normal:
- Média ($\mu$): 4,20
- Desvio-padrão ($\sigma$): 0,15

A variável aleatória contínua $X$ representa o pH medido em uma unidade de produto selecionada aleatoriamente em nosso lote de fabricação.

## 3. Questão 1 – Distribuição de Probabilidade

**Pergunta:** Escreva a distribuição de probabilidade de $X$ e identifique seus parâmetros.

### Formulação Estatística
A variável aleatória contínua $X$ segue uma distribuição normal com média $\mu = 4,20$ e desvio-padrão $\sigma = 0,15$:

$$X \sim N(\mu, \sigma^2)$$

$$X \sim N(4,20; 0,15^2)$$

### Identificação e Significado dos Parâmetros
- **Média ($\mu = 4,20$):** Indica o centro geométrico da distribuição de probabilidade. Em termos práticos, indica que, sob condições normais de produção, o valor médio esperado de pH para os lotes desse alimento é $4,20$.
- **Desvio-padrão ($\sigma = 0,15$):** Reflete a variabilidade inerente ou dispersão dos dados em torno do centro do processo. Um desvio-padrão de $0,15$ indica o quanto as unidades individuais tendem a variar sistematicamente do valor médio esperado.
- **Variância ($\sigma^2 = 0,0225$):** O quadrado do desvio-padrão, importante medida de dispersão associada à escala do processo.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import norm

# Definição das constantes fornecidas no enunciado
mu = 4.20
sigma = 0.15

print(f"Parâmetros definidos para as análises:")
print(f"Média (mu) = {mu:.2f}")
print(f"Desvio-padrão (sigma) = {sigma:.2f}")
print(f"Variância (sigma^2) = {sigma**2:.4f}")
```

## 4. Questão 2 – Probabilidade de pH $\le$ 4,00

**Pergunta:** Calcular $P(X \le 4,00)$.

### Notação Probabilística e Padronização
Para obtermos a probabilidade de uma unidade do produto apresentar pH igual ou menor do que $4,00$, realizamos a conversão da nossa escala original de pH para a distribuição Normal Padrão ($Z \sim N(0, 1)$).

O cálculo do escore padronizado $Z$ mede a quantos desvios-padrão o ponto de interesse está afastado da média:

$$Z = \frac{X - \mu}{\sigma}$$

Substituindo os valores do problema para $X = 4,00$:

$$Z = \frac{4,00 - 4,20}{0,15} = \frac{-0,20}{0,15} \approx -1,3333$$

A probabilidade desejada é, portanto:

$$P(X \le 4,00) = P(Z \le -1,3333)$$

Calcularemos essa área sob a curva normal acumulada utilizando a função `norm.cdf()` do Python.

```python
# Cálculo do Escore Z de forma programática
x_lim_inf = 4.00
z_score_2 = (x_lim_inf - mu) / sigma

# Cálculo da probabilidade acumulada P(X <= 4.00)
prob_q2 = norm.cdf(x_lim_inf, loc=mu, scale=sigma)

print(f"Cálculos Estatísticos da Questão 2:")
print(f"Escore Padronizado (Z) = {z_score_2:.4f}")
print(f"Probabilidade em valor decimal = {prob_q2:.6f}")
print(f"Probabilidade em valor percentual = {prob_q2 * 100:.4f}%")
```

### Interpretação do Resultado
O resultado de aproximadamente **$9,1211\%$** (ou $0,0912$ em escala decimal) de probabilidade significa que, em condições ideais de processo e controle estático estabelecido, cerca de **$9,12\%$ das unidades fabricadas apresentarão pH menor ou igual a $4,00$.**

No contexto do controle de qualidade, esse valor permite que a equipe industrial monitore e tome decisões preventivas. Dependendo da finalidade do alimento, se o pH muito baixo for prejudicial ou violar os parâmetros ideais de formulação, essa fração do lote estaria classificada como inconforme.

## 5. Questão 3 – Probabilidade entre 4,00 e 4,40

**Pergunta:** Calcular $P(4,00 \le X \le 4,40)$.

### Notação Probabilística
Para calcularmos a probabilidade de uma variável contínua estar em um intervalo fechado, calculamos a diferença das probabilidades acumuladas nos limites superior e inferior:

$$P(4,00 \le X \le 4,40) = P(X \le 4,40) - P(X \le 4,00)$$

Em termos de escore padronizado:
- Para $X_1 = 4,00 \implies Z_1 = -1,3333$
- Para $X_2 = 4,40 \implies Z_2 = \frac{4,40 - 4,20}{0,15} = \frac{0,20}{0,15} \approx +1,3333$

Logo:

$$P(-1,3333 \le Z \le 1,3333) = P(Z \le 1,3333) - P(Z \le -1,3333)$$

```python
# Limites do intervalo
x_inf = 4.00
x_sup = 4.40

# Cálculo das probabilidades individuais acumuladas
prob_acum_inf = norm.cdf(x_inf, loc=mu, scale=sigma)
prob_acum_sup = norm.cdf(x_sup, loc=mu, scale=sigma)

# Probabilidade no intervalo desejado
prob_q3 = prob_acum_sup - prob_acum_inf

print(f"Cálculos Estatísticos da Questão 3:")
print(f"P(X <= 4.40) = {prob_acum_sup:.6f}")
print(f"P(X <= 4.00) = {prob_acum_inf:.6f}")
print(f"P(4.00 <= X <= 4.40) em valor decimal = {prob_q3:.6f}")
print(f"P(4.00 <= X <= 4.40) em valor percentual = {prob_q3 * 100:.4f}%")
```

### Interpretação do Resultado
O valor encontrado de aproximadamente **$81,7578\%$** ($0,8176$ em decimal) demonstra que **$81,76\%$ das unidades produzidas possuem seu pH perfeitamente enquadrado na faixa recomendada entre $4,00$ e $4,40$.**

Em ciência de alimentos, se essa for a faixa ideal estabelecida de pH que conserva as qualidades sensoriais e microbiológicas do produto sem estragá-lo, o resultado indica um processo com bom nível de homogeneidade e centralização.

## 6. Questão 4 – Gráficos das Probabilidades

Nesta etapa, usaremos o `matplotlib.pyplot` para plotar dois gráficos individuais em alta definição e de cunho acadêmico.

```python
# Geração de valores no eixo X ao redor da média para desenhar a curva contínua
x_vals = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y_vals = norm.pdf(x_vals, loc=mu, scale=sigma)

# --- GRÁFICO 1: P(X <= 4.00) ---
plt.figure(figsize=(10, 6), dpi=100)
plt.plot(x_vals, y_vals, 'b-', label='Curva Normal', linewidth=2)

# Sombrear a área para X <= 4.00
x_fill1 = np.linspace(mu - 4 * sigma, 4.00, 500)
y_fill1 = norm.pdf(x_fill1, loc=mu, scale=sigma)
plt.fill_between(x_fill1, y_fill1, color='lightblue', alpha=0.6, label='P(X <= 4.00) = 9.12%')

# Linhas de referência
plt.axvline(x=mu, color='red', linestyle='--', label=f'Média (mu) = {mu:.2f}')
plt.axvline(x=4.00, color='darkorange', linestyle=':', linewidth=2, label='pH = 4.00')

# Customizações estéticas
plt.title('Distribuição de Probabilidade do pH: Região Crítica Inferior (pH <= 4.00)', fontsize=12, fontweight='bold')
plt.xlabel('Valores de pH', fontsize=10)
plt.ylabel('Densidade de Probabilidade', fontsize=10)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right', frameon=True)
plt.xlim(mu - 3.5 * sigma, mu + 3.5 * sigma)
plt.show()
```

```python
# --- GRÁFICO 2: P(4.00 <= X <= 4.40) ---
plt.figure(figsize=(10, 6), dpi=100)
plt.plot(x_vals, y_vals, 'b-', label='Curva Normal', linewidth=2)

# Sombrear a área entre 4.00 e 4.40
x_fill2 = np.linspace(4.00, 4.40, 500)
y_fill2 = norm.pdf(x_fill2, loc=mu, scale=sigma)
plt.fill_between(x_fill2, y_fill2, color='lightgreen', alpha=0.6, label='P(4.00 <= X <= 4.40) = 81.76%')

# Linhas de referência
plt.axvline(x=mu, color='red', linestyle='--', label=f'Média (mu) = {mu:.2f}')
plt.axvline(x=4.00, color='darkorange', linestyle=':', linewidth=2, label='pH = 4.00')
plt.axvline(x=4.40, color='purple', linestyle=':', linewidth=2, label='pH = 4.40')

# Customizações estéticas
plt.title('Distribuição de Probabilidade do pH: Faixa Ótima Recomendada (4.00 <= pH <= 4.40)', fontsize=12, fontweight='bold')
plt.xlabel('Valores de pH', fontsize=10)
plt.ylabel('Densidade de Probabilidade', fontsize=10)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right', frameon=True)
plt.xlim(mu - 3.5 * sigma, mu + 3.5 * sigma)
plt.show()
```

## 7. Questão 5 – Percentis 90, 95 e 99

**Pergunta:** Determinar os percentis $P_{90}$, $P_{95}$ e $P_{99}$.

### O que é o Percentil em Estatística?
Um percentil indica o valor abaixo do qual uma determinada porcentagem das observações da variável aleatória se enquadra. Por exemplo, o percentil 90 ($P_{90}$) é o valor de pH tal que a probabilidade acumulada sob a curva da normal seja de exatamente $90\%$ ou $0,90$. Para isso, utilizamos a função cumulativa inversa (também chamada de função quantil, em Python obtida via `norm.ppf()`).

```python
# Cálculo dos percentis com norm.ppf
x90 = norm.ppf(0.90, loc=mu, scale=sigma)
x95 = norm.ppf(0.95, loc=mu, scale=sigma)
x99 = norm.ppf(0.99, loc=mu, scale=sigma)

# Construção do dicionário e tabela usando Pandas
data_percentis = {
    'Percentil': ['90', '95', '99'],
    'Probabilidade acumulada': [0.90, 0.95, 0.99],
    'Valor de pH calculado': [f"{x90:.4f}", f"{x95:.4f}", f"{x99:.4f}"]
}

df_percentis = pd.DataFrame(data_percentis)
display(df_percentis)
```

Caso exista algum problema associado a microrganismos deteriorantes cujo crescimento seja favorecido por valores de pH menos ácidos, saber que $90\%$ da produção se mantém seguramente abaixo de $4,392$ é um forte indício de estabilidade físico-química.

## 9. Questão 7 – Percentil 99 e Comparação com o Percentil 90

### Interpretação do Percentil 99
O valor associado ao percentil 99 é aproximadamente **$4,549$**. Isso nos assegura que **$99\%$ das unidades do produto alimentício avaliadas exibem pH menor ou igual a $4,549$**, restando uma fração insignificante de $1\%$ das unidades produzidas com pH excedendo essa marca.

### Comparação Técnica entre os Percentis $P_{90}$ e $P_{99}$
- **Diferença de Valores:** O pH associado ao $P_{99}$ ($4,549$) é significativamente maior do que o pH de $P_{90}$ ($4,392$).
- **Distância em relação à média ($\mu = 4,20$):** O $P_{99}$ está posicionado a uma distância superior em termos de desvios-padrão da média se comparado ao $P_{90}$. Enquanto o $P_{90}$ localiza-se a aproximadamente $+1,28$ desvios-padrão de distância, o $P_{99}$ afasta-se a $+2,33$ desvios-padrão.
- **Significado Estatístico:** Para passar do escopo de $90\%$ para $99\%$ de probabilidade acumulada sob a cauda direita, precisamos incorporar mais $9\%$ da população total do produto. Como a densidade de probabilidade decresce vertiginosamente à medida que caminhamos em direção às caudas (extremos) da curva normal, é preciso expandir o intervalo do pH por uma margem proporcionalmente maior na escala física de medição para capturar esses casos mais extremos e raros.
