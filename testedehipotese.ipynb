# Resolução dos Exercícios Selecionados

Abaixo apresentamos os códigos e análises para as três questões de interesse: **Média com Variância Desconhecida**, **Variância Populacional** e **Proporção Populacional**.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Configurações gerais dos gráficos
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
```

## Resolução Detalhada dos Exercícios Selecionados

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Definição do nível de significância
alpha = 0.05

# =========================================================================
# QUESTÃO 2: Média com Variância Desconhecida (Teor de Proteína)
# =========================================================================
print("=== QUESTÃO 2 ===")
mu0_q2 = 8.0
n_q2 = 16
mean_q2 = 8.4
s_q2 = 0.8

# Estatística de Teste t
t_obs_q2 = (mean_q2 - mu0_q2) / (s_q2 / np.sqrt(n_q2))
df_q2 = n_q2 - 1
# Teste unicaudal superior
t_crit_q2 = stats.t.ppf(1 - alpha, df_q2)
p_val_q2 = 1 - stats.t.cdf(t_obs_q2, df_q2)

print(f"t observado: {t_obs_q2:.4f}")
print(f"t crítico: {t_crit_q2:.4f}")
print(f"Valor-p: {p_val_q2:.4f}")
print(f"Decisão: Rejeitar H0? {'Sim' if p_val_q2 < alpha else 'Não'}\n")

# Gráfico Q2
x_q2 = np.linspace(-4, 4, 1000)
y_q2 = stats.t.pdf(x_q2, df_q2)

plt.figure(figsize=(10, 4))
plt.plot(x_q2, y_q2, label=f't-Student (df={df_q2})', color='blue')
plt.axvline(t_crit_q2, color='red', linestyle='--', label=f't Crítico = {t_crit_q2:.3f}')
plt.axvline(t_obs_q2, color='green', linestyle='-', label=f't Obs = {t_obs_q2:.3f}')

# Região Crítica
x_crit_q2 = np.linspace(t_crit_q2, 4, 100)
plt.fill_between(x_crit_q2, stats.t.pdf(x_crit_q2, df_q2), color='red', alpha=0.3, label='Região de Rejeição (5%)')

plt.title('Distribuição t-Student: Teste de Média (Teor de Proteína)')
plt.xlabel('Valor t')
plt.ylabel('Densidade de Probabilidade')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# =========================================================================
# QUESTÃO 4: Variância Populacional (pH do Produto Fermentado)
# =========================================================================
print("=== QUESTÃO 4 ===")
var0_q4 = 0.04
n_q4 = 20
s2_q4 = 0.064
df_q4 = n_q4 - 1

# Estatística de Teste Qui-Quadrado
chi_obs_q4 = (n_q4 - 1) * s2_q4 / var0_q4

# Teste bicaudal
chi_crit_inf_q4 = stats.chi2.ppf(alpha / 2, df_q4)
chi_crit_sup_q4 = stats.chi2.ppf(1 - alpha / 2, df_q4)

# Valor-p para teste bicaudal
p_val_q4 = 2 * min(stats.chi2.cdf(chi_obs_q4, df_q4), 1 - stats.chi2.cdf(chi_obs_q4, df_q4))

print(f"Qui-quadrado observado: {chi_obs_q4:.4f}")
print(f"Valores Críticos: [{chi_crit_inf_q4:.4f}, {chi_crit_sup_q4:.4f}]")
print(f"Valor-p: {p_val_q4:.4f}")
print(f"Decisão: Rejeitar H0? {'Sim' if p_val_q4 < alpha else 'Não'}\n")

# Gráfico Q4
x_q4 = np.linspace(0, 45, 1000)
y_q4 = stats.chi2.pdf(x_q4, df_q4)

plt.figure(figsize=(10, 4))
plt.plot(x_q4, y_q4, label=f'Qui-Quadrado (df={df_q4})', color='purple')
plt.axvline(chi_crit_inf_q4, color='red', linestyle='--', label=f'Lim Inf = {chi_crit_inf_q4:.3f}')
plt.axvline(chi_crit_sup_q4, color='red', linestyle='--', label=f'Lim Sup = {chi_crit_sup_q4:.3f}')
plt.axvline(chi_obs_q4, color='green', linestyle='-', label=f'Qui-2 Obs = {chi_obs_q4:.3f}')

# Regiões Críticas
x_crit_inf_q4 = np.linspace(0, chi_crit_inf_q4, 100)
x_crit_sup_q4 = np.linspace(chi_crit_sup_q4, 45, 100)
plt.fill_between(x_crit_inf_q4, stats.chi2.pdf(x_crit_inf_q4, df_q4), color='red', alpha=0.3, label='Região Rejeição Inf')
plt.fill_between(x_crit_sup_q4, stats.chi2.pdf(x_crit_sup_q4, df_q4), color='red', alpha=0.3, label='Região Rejeição Sup')

plt.title('Distribuição Qui-Quadrado: Teste de Variância (pH)')
plt.xlabel('Valor Qui-quadrado')
plt.ylabel('Densidade de Probabilidade')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# =========================================================================
# QUESTÃO 5: Proporção Populacional (Não conformidade microbiológica)
# =========================================================================
print("=== QUESTÃO 5 ===")
p0_q5 = 0.05
n_q5 = 200
x_q5 = 16
p_hat_q5 = x_q5 / n_q5

# Estatística de Teste Z
z_obs_q5 = (p_hat_q5 - p0_q5) / np.sqrt((p0_q5 * (1 - p0_q5)) / n_q5)

# Teste unicaudal superior
z_crit_q5 = stats.norm.ppf(1 - alpha)
p_val_q5 = 1 - stats.norm.cdf(z_obs_q5)

print(f"Proporção amostral (p_hat): {p_hat_q5:.4f} ({p_hat_q5*100:.1f}%)")
print(f"Z observado: {z_obs_q5:.4f}")
print(f"Z crítico: {z_crit_q5:.4f}")
print(f"Valor-p: {p_val_q5:.4f}")
print(f"Decisão: Rejeitar H0? {'Sim' if p_val_q5 < alpha else 'Não'}\n")

# Gráfico Q5
x_q5_plot = np.linspace(-4, 4, 1000)
y_q5_plot = stats.norm.pdf(x_q5_plot)

plt.figure(figsize=(10, 4))
plt.plot(x_q5_plot, y_q5_plot, label='Normal Padrão N(0,1)', color='orange')
plt.axvline(z_crit_q5, color='red', linestyle='--', label=f'Z Crítico = {z_crit_q5:.3f}')
plt.axvline(z_obs_q5, color='green', linestyle='-', label=f'Z Obs = {z_obs_q5:.3f}')

# Região Crítica
x_crit_q5 = np.linspace(z_crit_q5, 4, 100)
plt.fill_between(x_crit_q5, stats.norm.pdf(x_crit_q5), color='red', alpha=0.3, label='Região de Rejeição (5%)')

plt.title('Distribuição Normal Padrão: Teste de Proporção (Microbiológico)')
plt.xlabel('Valor Z')
plt.ylabel('Densidade de Probabilidade')
plt.legend()
plt.tight_layout()
plt.show()
```
