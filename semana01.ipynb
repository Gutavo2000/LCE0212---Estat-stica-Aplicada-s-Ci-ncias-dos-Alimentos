```python
import pandas as pd

file_path = '/content/dados.xlsx'

try:
    df = pd.read_excel(file_path)
    print("Dados importados com sucesso! Todo o DataFrame:")
    display(df)
except FileNotFoundError:
    print(f"Erro: O arquivo '{file_path}' não foi encontrado. Verifique o caminho do arquivo.")
except Exception as e:
    print(f"Ocorreu um erro ao ler o arquivo Excel: {e}")
```

```python
print(df.columns)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fruit_preference_column = 'Qual dessas frutas você prefere para suco natural?'

# Count the occurrences of each fruit preference
preference_counts = df[fruit_preference_column].value_counts()

# Create the bar chart
plt.figure(figsize=(10, 6))
sns.barplot(x=preference_counts.index, y=preference_counts.values, hue=preference_counts.index, palette='viridis', legend=False)
plt.title('Preferência de Frutas para Suco Natural')
plt.xlabel('Fruta')
plt.ylabel('Número de Respostas')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

# Tipos de variaveis

# Qualitativas nominal
• Cor dos olhos: azul, verde, castanho

• Estado civil: solteiro, casado, divorciado

• Tipo sanguíneo: A, B, AB, O

# Qualitativas ordinal
• Grau de satisfação: insatisfeito, neutro, satisfeito

• Escolaridade: ensino fundamental, médio, superior

• Classificação: ruim, regular, bom, excelente

# Quantitativas Discreta
• Número de filhos: 0, 1, 2, 3...

• Quantidade de alunos em uma sala

• Número de carros de uma família

# Quantitativas Continua
• Altura: 1,65 m; 1,72 m

• Peso: 60,5 kg

• Temperatura: 36,7 °C
